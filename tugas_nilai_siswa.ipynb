```python
import pandas as pd
df = pd.read_csv('StudentsPerformance.csv')
df
```

```python
df['Total Score'] = df['math score'] + df['reading score'] + df['writing score']
```

```python
df['math score'].mean()
```

```python
df['Total Score'].mean()
```

```python
#sebuah perusahaan mencari orang yang mempunyai bachelor degree atau masters degree, mempunyai skor math lebih dari 80, dan punya total skor lebih 210
cari= df[df['Total Score']>=210][df['math score']>=80]['parental level of education'].value_counts()
print (cari["master's degree"])
print (cari["bachelor's degree"])
```
